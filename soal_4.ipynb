```python
import math

# Menentukan titik pusat dan titik yang akan dihampiri
a = 25
x = 26
h = x - a

# Menentukan nilai eksak sebagai pembanding
nilai_eksak = math.sqrt(x)

# Menyimpan nilai koefisien Taylor dari orde 0 sampai 4
koefisien = [5, 0.1, -0.001, 0.00002, -0.0000002]

# Menyimpan hasil hampiran untuk setiap orde
hasil = []

# Menghitung polinomial Taylor dari P0 sampai P4
for n in range(5):
    # Menjumlahkan setiap suku polinomial sampai orde n
    pn = sum(koefisien[k] * h**k for k in range(n + 1))

    # Menghitung selisih absolut terhadap nilai eksak
    galat = abs(nilai_eksak - pn)

    # Menyimpan hasil dan galat setiap orde
    hasil.append((n, pn, galat))

# Menampilkan nilai pembanding dengan presisi penuh
print(f"Nilai eksak sqrt(26) = {nilai_eksak:.12f}")
print()

# Menampilkan hasil setiap orde dan galat absolutnya
print("Orde | Hasil hampiran | Galat absolut")
for n, pn, galat in hasil:
    print(f"P{n}   | {pn:.12f} | {galat:.12e}")

# Menentukan batas galat yang diminta soal
batas_galat = 1e-6

# Mencari orde pertama yang galatnya lebih kecil dari batas
for n, pn, galat in hasil:
    if galat < batas_galat:
        print()
        print(f"Orde minimal = {n}")
        print(f"Hasil hampiran = {pn:.12f}")
        print(f"Galat absolut = {galat:.12e}")
        break
else:
    print("Belum ada orde yang memenuhi batas galat.")
```
