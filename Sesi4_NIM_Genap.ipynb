```python
import math

# Fungsi persamaan nonlinear
def f(t):
    return 800 * math.exp(0.3 * t) - 2000 - 400 * t

# Interval awal
a = 4.0
b = 6.0

# Toleransi dan iterasi maksimum
tol = 0.0001
maks_iterasi = 100

# Memastikan interval mengapit akar merupakan langkah untuk mengetahui apakah suatu persamaan memiliki akar pada rentang nilai tertentu.
if f(a) * f(b) >= 0:
    print("Interval tidak memenuhi syarat Regula Falsi.")
else:
    print("METODE REGULA FALSI")
    print("-" * 85)
    print(f"{'Iterasi':<10}{'a':>12}{'b':>12}"
          f"{'xr':>12}{'f(xr)':>15}{'Error':>15}")
    print("-" * 85)

    xr_lama = None

    for i in range(1, maks_iterasi + 1):
        fa = f(a)
        fb = f(b)

        # Rumus Regula Falsi
        xr = (a * fb - b * fa) / (fb - fa)
        fxr = f(xr)

        # Menghitung error absolut
        if xr_lama is None:
            error = float("inf")
        else:
            error = abs(xr - xr_lama)

        print(f"{i:<10}{a:>12.4f}{b:>12.4f}"
              f"{xr:>12.4f}{fxr:>15.4f}", end="")

        if xr_lama is None:
            print(f"{'-':>15}")
        else:
            print(f"{error:>15.4f}")

        # Berhenti jika fungsi mendekati nol
        if abs(fxr) < tol:
            break

        # Memperbarui interval
        if fa * fxr < 0:
            b = xr
        else:
            a = xr

        xr_lama = xr

    print("-" * 85)
    print(f"Akar persamaan = {xr:.4f}")
    print(f"f(xr) = {f(xr):.4f}")
    print(f"Jumlah iterasi = {i}")
```
